```python
import pandas as pd
cosmo_df = pd.DataFrame({
    'Cosmology': cosmo_names,
    'File_name': file_names},
    index=cosmo_tags
)
```

```python
cosmo_df.loc['c000']['Cosmology']
```

```python
from astropy.cosmology import w0waCDM
import abacusnbody.metadata

meta_time_indp = abacusnbody.metadata.get_meta(cosmo_df.loc['c000']['Cosmology'])
meta = abacusnbody.metadata.get_meta(cosmo_df.loc['c000']['Cosmology'],redshift=0.1)

cosmo = w0waCDM(H0=meta_time_indp['H0'],
                Om0=meta_time_indp['Omega_M'], 
                Ode0=meta_time_indp['Omega_DE'], 
                Ob0=meta_time_indp['omega_b'],
                w0=meta_time_indp['w0'], 
                wa=meta_time_indp['wa'],
                Neff=meta_time_indp['N_ur']+meta_time_indp['N_ncdm'],
                Tcmb0=2.7255)
print(cosmo)
```

# Extract Abacus cosmologies from csv

```python
cosmo_params_df = pd.read_csv('cosmologies.csv')

len(cosmo_params_df), len(cosmo_tags)
```

```python
cosmo_numbers = [int(tag[1:]) for tag in cosmo_tags]

cosmo_params_df['number'] = cosmo_params_df['root'].str.extract(r'(\d+)$').astype(int)

filtered_cosmo = cosmo_params_df[cosmo_params_df['number'].isin(cosmo_numbers)]

filtered_cosmo = filtered_cosmo.drop(columns=['number'])

len(filtered_cosmo)
```

```python
cosmo_tags.sort()
filtered_cosmo.index = cosmo_tags
filtered_cosmo.index.name = 'cosmo'
filtered_cosmo.drop(columns=['root'], inplace=True)
filtered_cosmo
```

```python
latex_columns = {
    'root': 'Cosmology',
    'notes': 'Notes',
    'omega_b' : r'$\Omega_b$',
    'omega_cdm' : r'$\Omega_c$',
    'h': r'$h$',
    'A_s': r'$A_s$',
    'n_s': r'$n_s$',
    'alpha_s': r'$\alpha_s$',
    'N_ur': r'$N_{ur}$',
    'N_ncdm': r'$N_{ncdm}$',
    'omega_ncdm': r'$\Omega_{ncdm}$',
    'w0_fld': r'$w_0$',
    'wa_fld': r'$w_a$',
    'sigma8_m': r'$\sigma_8$',
    'sigma8_cb': r'$\sigma_{8,cb}$',
}

filtered_cosmo_renamed = filtered_cosmo.rename(columns=latex_columns)

filtered_cosmo_renamed.to_csv('filtered_cosmologies_latex_final.csv')
```
